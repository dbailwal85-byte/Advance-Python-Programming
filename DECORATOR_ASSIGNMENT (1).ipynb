# DECORATOR

Create a simple decorator that adds extra print statements before and after a function execution

```python
# Simple decorator

def my_decorator(func):
    def wrapper():
        print("Function execution started")
        
        func()
        
        print("Function execution finished")
    
    return wrapper


# Using the decorator
@my_decorator
def greet():
    print("Hello, World!")


# Calling the function
greet()
```

Create a decorator that wraps a function and prints a custom message before calling the original function

```python
# Decorator with a custom message

def message_decorator(func):
    def wrapper():
        print("Custom Message: Function is about to run!")
        
        func()
    
    return wrapper


# Original function
@message_decorator
def say_hello():
    print("Hello!")

# Calling the function
say_hello()
```

Create a decorator that takes a function with two arguments and logs the function name along with its parameters

```python
# Decorator for logging function name and parameters

def log_decorator(func):
    def wrapper(a, b):
        print(f"Calling function: {func.__name__}")
        print(f"Arguments: a = {a}, b = {b}")
        
        result = func(a, b)
        
        return result
    return wrapper


# Function with two arguments
@log_decorator
def add(x, y):
    return x + y


# Calling the function
result = add(10, 20)

print("Result:", result)
```

Create a decorator to measure execution time of a function that takes no arguments

```python
import time

# Decorator to measure execution time
def time_decorator(func):
    def wrapper():
        start_time = time.time()
        
        func()
        
        end_time = time.time()
        
        print(f"Execution Time: {end_time - start_time:.4f} seconds")
    
    return wrapper


# Function with no arguments
@time_decorator
def task():
    print("Task is running...")
    time.sleep(2)   # Simulate delay


# Calling the function
task()
```

Create a decorator that checks if the input user is "admin" before allowing function execution

```python
# Decorator to check admin access

def admin_only(func):
    def wrapper(user):
        if user == "admin":
            return func(user)
        else:
            print("Access Denied! Only admin can execute this function.")
    
    return wrapper


# Function to protect
@admin_only
def delete_data(user):
    print(f"Welcome {user}! Data deleted successfully.")


# Testing
delete_data("admin")
delete_data("guest")
```

Create a decorator that caches results of a function with one integer argument to avoid repeated calculations

```python
# Decorator for caching function results

def cache_decorator(func):
    cache = {}

    def wrapper(n):
        if n in cache:
            print(f"Fetching result from cache for {n}")
            return cache[n]

        print(f"Calculating result for {n}")
        result = func(n)
        cache[n] = result
        return result

    return wrapper


# Function with one integer argument
@cache_decorator
def square(num):
    return num * num


# Testing
print(square(5))
print(square(5))   # Cached result
print(square(10))
print(square(10))  # Cached result
```

Create a decorator that works with functions having arguments using *args and **kwargs and prints messages before and after execution

```python
# Decorator using *args and **kwargs

def message_decorator(func):
    def wrapper(*args, **kwargs):
        print("Function execution started")
        
        result = func(*args, **kwargs)
        
        print("Function execution finished")
        
        return result
    
    return wrapper


# Function with arguments
@message_decorator
def multiply(a, b):
    print(f"Multiplying {a} and {b}")
    return a * b


# Calling the function
result = multiply(4, 5)

print("Result:", result)
```

Create a decorator that modifies the return value of a function by doubling the result

```python
# Decorator to double the return value

def double_result(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result * 2
    
    return wrapper


# Function example
@double_result
def add(a, b):
    return a + b


# Calling the function
print(add(5, 10))
```
