# Iterators

Write a program to create an iterator from a list and print all elements using a while loop and next()

```python
# Create a list
numbers = [10, 20, 30, 40, 50]

# Create an iterator from the list
iterator = iter(numbers)

# Print elements using while loop and next()
while True:
    try:
        print(next(iterator))
    except StopIteration:
        break
```

Write a program to demonstrate what happens when next() is called more times than the number of elements in an iterator

```python
# Create a list
numbers = [1, 2, 3]

# Create an iterator
it = iter(numbers)

# Calling next() more times than elements
print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3

# No more elements left
print(next(it))  # Error
```

Write a program to iterate over an iterator using a for loop and print all elements

```python
# Create a list
numbers = [10, 20, 30, 40, 50]

# Create an iterator from the list
my_iterator = iter(numbers)

# Iterate over the iterator using a for loop
for item in my_iterator:
    print(item)
```

Write a program to create an iterator using range() and print values step by step using next()

```python
# Create an iterator using range()
numbers = iter(range(1, 6))

# Print values step by step using next()
print(next(numbers))
print(next(numbers))
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

Write a program to manually traverse a list using iter() and next() and handle StopIteration using try-except

```python
# Create a list
fruits = ["Apple", "Banana", "Mango"]

# Create an iterator from the list
my_iterator = iter(fruits)

# Traverse the list manually using next()
try:
    print(next(my_iterator))
    print(next(my_iterator))
    print(next(my_iterator))
    
    # This will raise StopIteration
    print(next(my_iterator))

except StopIteration:
    print("End of iterator reached")
```

# Generator

Write a generator function using yield that returns multiple values and print them using next()

```python
# Generator function
def my_generator():
    yield "Python"
    yield "Java"
    yield "C++"

# Create generator object
gen = my_generator()

# Print values using next()
print(next(gen))
print(next(gen))
print(next(gen))
```

Write a program to iterate over a generator using a for loop and display all values

```python
# Generator function
def numbers():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5

# Iterate over the generator using a for loop
for value in numbers():
    print(value)
```

Write a generator expression to generate square values and print first few elements using next()

```python
# Generator expression for square values
squares = (x * x for x in range(1, 11))

# Print first few elements using next()
print(next(squares))
print(next(squares))
print(next(squares))
print(next(squares))
print(next(squares))
```

Write a generator function to generate Fibonacci series and print the output as a list

```python
# Generator function for Fibonacci series
def fibonacci(n):
    a, b = 0, 1
    
    for i in range(n):
        yield a
        a, b = b, a + b

# Convert generator output into a list
fib_list = list(fibonacci(10))

# Print the Fibonacci series
print(fib_list)
```

Write a program to demonstrate generator methods like next() and send() to control the flow of execution

```python
# Generator function
def my_generator():
    print("Generator started")
    
    value = yield "First Value"
    print("Received:", value)
    
    value = yield "Second Value"
    print("Received:", value)
    
    yield "Third Value"

# Create generator object
gen = my_generator()

# Start generator using next()
print(next(gen))

# Send value to generator
print(gen.send("Hello"))

# Send another value
print(gen.send("Python"))
```
